```python
# Pipeline B: For normal floats, binary flags, and balanced ints (Impute -> Scale)
normal_num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy="median")),
    ('scaler', StandardScaler())
])

# Pipeline C: For clean categorical features like TLD (Impute -> One-Hot)
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="constant", fill_value="None")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])


# --- 3. Tie everything together into a Preprocessor ---
preprocessor = ColumnTransformer(
    transformers=[
        ('skewed', skewed_pipeline, skewed_cols),
        ('normal_num', normal_num_pipeline, normal_num_cols),
        ('cat', categorical_pipeline, cat_cols)
    ],
    remainder='drop' # This safely drops raw 'URL', 'Domain', and 'FILENAME' columns automatically!
)
```
